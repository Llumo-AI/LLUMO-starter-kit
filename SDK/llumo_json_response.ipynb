# **📘 Notebook Overview**

---

This notebook demonstrates how to **generate, test, and evaluate responses** from large language models (LLMs) like **GPT-4**, using **relevant contextual information** (e.g., via Retrieval-Augmented Generation).

🧪 The evaluation focuses on multiple key performance indicators (KPIs), including:

- 🎯 **Response Correctness**  
- 🧩 **Response Completeness**  
- 🧠 **Response Bias**  
- ☣️ **Response Harmfulness**

---

🔍 The process uses the `LlumoClient` API, which provides **automated, structured assessments** of model responses to ensure high quality and safety.

#**🔑 Setup Llumo API key**


```python
import os

# Set your Llumo API Key
os.environ["LLUMO_API_KEY"] = "Enter Your llumo API Key"

llumo_key = os.getenv("LLUMO_API_KEY")
```

# **✅ Sample Data for Evaluating AI Outputs**


```python
# This data simulates user queries, and supporting context, output.
data = [
    {
        "query": "What's the capital of France?",
        "context": "Paris is the capital city of France, known for its historical landmarks like the Eiffel Tower and its role in politics, culture, and art.",
        "output": "The capital of France is Paris."
    },
    {
        "query": "Summarize the mission of OpenAI.",
        "context": "OpenAI is a research company focused on ensuring that artificial general intelligence (AGI) benefits all of humanity through safety, transparency, and broad cooperation.",
        "output": "OpenAI is dedicated to developing artificial general intelligence (AGI) in a safe and beneficial way for all of humanity, emphasizing collaboration, transparency, and safety."
    },
    {
        "query": "Who won the 2022 FIFA World Cup?",
        "context": "The 2022 FIFA World Cup was held in Qatar, and Argentina emerged victorious after defeating France in a dramatic penalty shootout.",
        "output": "Argentina won the 2022 FIFA World Cup after defeating France in a thrilling penalty shootout."
    }
]
```

#**🔍 LLumo Evaluation: Evaluate Model Responses Using LlumoClient**


```python

# Import the evaluation client from Llumo SDK
from llumo import LlumoClient

# Initialize the LlumoClient with your API key
client = LlumoClient(api_key = llumo_key)  # Replace with actual API key

# Evaluate the Data using selected evaluation KPIs
resultdf = client.evaluateMultiple(
    data = data,  # Input Data
    evals = ["Response Completeness", "Response Correctness", "Response Bias", "Response Harmfulness"],  # Selected evaluation KPIs
    prompt_template = "Give answer to the given query: {{query}}, using the given context: {{context}}.",  # Prompt used for generation
    createExperiment = False,   # Set to True to save results as an experiment on the Llumo platform. If False, returns results as a DataFrame or A Python Dict.
    getDataFrame = True, # Return result as a DataFrame (True) or dictionary (False)
)
```
