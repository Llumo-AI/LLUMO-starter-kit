# **Objective: Demonstrating Multiple LLM Models at the Same Time with Llumo with Evaluation**

This notebook demonstrates how we can run the same prompt across multiple language models simultaneously and compare the results using **Llumo**. We will also show how **evaluation** works when we set `toEvaluate=True` , `toEvaluate=False` also `createExperimet=True`, `createExperiment=False`.


## **1. Install and Import Required Libraries**

We begin by importing the required libraries for this task, including **pandas** and **LlumoClient**. We will also import `AVAILABLEMODELS` to specify the models we wish to evaluate.


```python
!pip install llumo
```

```python
import pandas as pd
from llumo import LlumoClient, AVAILABLEMODELS
```

## **2. Initialize LlumoClient**
Here, we initialize the LlumoClient using our API key, which is used to interact with the Llumo API for model evaluation.

```python
client = LlumoClient(api_key="Enter Your own api key")
```

## **3. Define Dataset and Models**
Here, we define the dataset containing the topics that we want the models to explain. We will use a dataset of 2 topics: "Quantum Physics" and "Politics".

We also specify the models to be used in the evaluation. We are using the CHAT_BISON and GOOGLE_15_FLASH models from Llumo's available models.

```python
dataset = {
    "topic": ["Quantum Physics", "Politics"]
}

models = [
    AVAILABLEMODELS.CHAT_BISON,
    AVAILABLEMODELS.GOOGLE_15_FLASH,
]

api_key = "Your Api key"# Enter Your own api key according to the model you choose
```

## **4. Define Prompt Template**
The prompt template defines how the input (in this case, the topic) should be transformed into a query. We are asking the models to explain the topic in a very simple way.

```python
prompt_template = ["Explain {{topic}} to me like I'm a 5 year old. in 5 words"]
```

## **5. Run Evaluation with toEvaluate=False**
Now, we will run the same prompt across the models, but this time, evaluation is disabled by setting toEvaluate=False. This will return the raw results without the evaluation metrics.
By default toEvaluate is set to False

```python
res = client.run_sweep(templates=prompt_template, dataset=dataset, model_aliases=models, apiKey=api_key, eval=[""],toEvaluate=False,createExperiment=False)
res
```

## **6. Run Evaluation with toEvaluate=True**
In this section, we will evaluate the models with evaluation enabled by setting toEvaluate=True. This will allow Llumo to assess the responses based on "Response Correctness" and "Response Completeness".

```python
res = client.run_sweep(templates=prompt_template, dataset=dataset, model_aliases=models, apiKey=api_key, eval=["Response Correctness","Response Completeness"],toEvaluate=True,createExperiment=False)
res
```
