###**🔑 2️⃣ API Keys Setup (Securely) 🛡️**

```python
import os

# Store your API keys securely
os.environ['OPENAI_API_KEY'] = "🔑 Replace with your OpenAI key"
os.environ['LLUMO_API_KEY'] = "🔑 Replace with your Llumo key"

# Retrieve keys
openai_key = os.getenv("OPENAI_API_KEY")
llumo_key = os.getenv("LLUMO_API_KEY")
```

###**3️⃣Sample Data 📂**

```python
# Example data with input queries
data = [
    {"query": "What's the capital of France?"},
    {"query": "Explain the mission of OpenAI."},
    {"query": "Who won the 2022 FIFA World Cup?"}
]
```

###**🤖 4️⃣ Generate Output using LangChain + OpenAI 🧠**

```python
from langchain_openai import ChatOpenAI
from langchain.schema import HumanMessage

# Initialize LangChain OpenAI model
llm = ChatOpenAI(openai_api_key=openai_key, model="gpt-3.5-turbo")

# Generate responses for each query and format for Llumo evaluation
results = []
for item in data:
    query = item["query"]
    response = llm([HumanMessage(content=query)]).content
    results.append({
        "query": query,
        "output": response
    })
```


results  # 🔍 Display the final data structure for Llumo evaluation


###**📊 5️⃣ Evaluate the Results using Llumo ✅**

```python
from llumo import LlumoClient

# Initialize Llumo client with API key
client = LlumoClient(api_key=llumo_key)

# Evaluate responses on selected KPIs
resultDf = client.evaluateMultiple(
    data=results,
    evals=["Response Sentiment", "Response Bias", "Response Harmfulness"],
    createExperiment=False,   # Just return results, no experiment creation
    getDataFrame=True         # Return as DataFrame for easy viewing
)

```
