### **📚 Import required libraries**


```python
import os
from llumo import LlumoClient, AVAILABLEMODELS
import random
import pandas as pd
```

###**📦 3️⃣ Prepare Sample Data (Based on How Successful People Think)**

```python
# 📝 Sample data related to created data stream
data = [
    {"query": "How do successful people think differently?"},
    {"query": "What are the daily habits of successful people?"},
    {"query": "How to maintain a positive mindset like successful people?"}
]
```

###**🤖 4️⃣ Configure Models to be Used for Generation**

```python
# 🧠 Choose LLM models for generation
models = [
    AVAILABLEMODELS.GPT_4o,   # Latest OpenAI GPT-4o model
    AVAILABLEMODELS.GPT_4     # OpenAI GPT-4
]
```

**🚀 5️⃣ Generate Outputs and Run Evaluations**


```python
# Initialize Llumo Client with API key
client = LlumoClient(api_key=LLUMO_API_KEY)

# 🎯 Run generation + evaluation on the data using llumo.ai
evalDf = client.ragSweep(
    data=data, # input data with a 'query' column
    streamName="Replace with your data stream name here", # data stream name
    queryColName="Replace with query column name", # name of the query column. Keep it 'query' to avoid any error during evaluation
    evals=["Redundancy Reduction", "Input Bias", "Context Utilization"], # list of eval metrics
    modelAliases=models, # list of models
    generateOutput=True, # Generate output using the `query` and fetched `context` when 'True'
    toEvaluate=True, # Perform evaluation when 'True'
    apiKey=OPENAI_API_KEY, # open api key
   createExperiment = False,   # Set to True to save results as an experiment on the Llumo platform. If False, returns results as a DataFrame or A Python Dict. - Optional
    getDataFrame = True, # Return result as a DataFrame (True) or dictionary (False) - Optional
 )
```
