# 🔐 **Step 2: Load Llumo API Key Securely**


```python
from google.colab import userdata

# Ensure you have stored your Llumo API key using:
# `userdata.set('LLUMO_API_KEY', 'your_actual_key')` in a previous cell or session

api_key = userdata.get("LLUMO_API_KEY")

# Validate the key
if not api_key:
    raise ValueError("API key not found. Please set your LLUMO_API_KEY in Colab userdata.")
```

# 📦 **Step 3: Import required libraries**


```python

import pandas as pd
from llumo import LlumoClient
from google.colab import userdata
```

#📋  **Step 4: Importing the data file**

```python
# upload your data file using pandas

file_name = "your_excel_file_here"
df = pd.read_excel(file_name)
```

# **🚀 Step 5: Fetch data from the data stream**

- start the searching by specifying the stream name.
-  you can find the strean name after logging into https://llumo.ai/
- the dataframe should have a query column - based on which the searching will be performed.

```python

# Initialize LlumoClient
client = LlumoClient(api_key=api_key)

# start fetching
output_df = client.runDataStream(
    dataframe= df,
    streamName = "datastream_name_here",
    queryColName = "query_column_name_here"
)
```
