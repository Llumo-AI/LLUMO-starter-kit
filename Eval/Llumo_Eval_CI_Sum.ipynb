This notebook evaluates the quality of LLM-generated responses based on key metrics such as confidence, relevance, and accuracy etc. The evaluation is performed using the LLUMO SDK.

Step 1: Import Required Libraries

````python
import requests
import json
import time
import pandas as pd
from google.colab import userdata
````

Step 2: Define API Configuration


````python
# Replace with a valid API key
LLUMO_API_KEY = "Enter Your Llumo AI Key Here"
````

Step 3: Load and Prepare Dataset


```
We begin by loading a dataset containing:

query: The input question or query.o
context: The context retrieved for the query.
groundTruth: The expected correct response.
output: The LLM-generated response.
```






````python
# Load dataset from an Excel file
# Ensure the dataset contains required columns: 'query', 'context', 'groundTruth', 'output'
data = pd.read_excel("dataset.xlsx", usecols=["query", "context", "groundTruth", "output"])
data = data[:5]

# list of evaluation metrics
metrics = ["Response Correctness","Input Relevancy", "Response Toxicity"]
````

Step 4: Process the dataframe and evaluate


````python
from llumo import LlumoClient
client = LlumoClient(api_key=LLUMO_API_KEY)
result_df = client.evaluate(data,evals = metrics,prompt_template="Provide answer for the given query: {{query}},using the knowledgebase : {{context}}")
````
