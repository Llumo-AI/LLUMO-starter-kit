This notebook generates responses for a given prompt using multiple LLM models (GPT-3.5-Turbo, Gemini-Pro, and Text-Bison) via the LLUMO API. The results are stored in a structured Excel sheet.

Step 1: Import Required Libraries

```python
import requests
import pandas as pd
from google.colab import userdata
```

Step 2: Define API Configuration

```python

LLUMO_API_KEY = userdata.get("LLUMO_API_KEY")
# API Authorization Header (Replace with a valid API key)
headers = {
    "Authorization": f"Bearer {LLUMO_API_KEY}"
}

# API Endpoint
LLUMO_ENDPOINT = "https://model-api.llumo.ai/analytics/get-single-output-analytics"
```

Step 3: Define Available LLM Models

```python
# List of LLM models to use for generating responses
llm_models = ["gpt-3.5-turbo", "gemini-pro", "text-bison"]
```

Step 4: Define Prompts for Evaluation

```python
import pandas as pd

# Define different prompt styles to generate child-friendly explanations.
prompt_templates = [
    "Explain {{topic}} in a way that a 10-year-old can understand.",
    "Give a simple and fun explanation of {{topic}} with an example.",
    "Break down {{topic}} into very easy steps for a child."
]

# Define topics for which explanations will be generated.
topics = ["Artificial Intelligence", "Photosynthesis", "Black Holes"]

# Initialize an empty DataFrame to store prompts and model outputs.
data = pd.DataFrame(columns=["Prompt"] + [f"output_{model}" for model in llm_models])
data
```

Step 5: Generating the LLM response.

```python
import json
# Iterate through prompt templates to generate explanations.
for indx in range(len(prompt_templates)):
    prompt = prompt_templates[indx]
    topic = topics[indx]  # Extract topic for API input
    finalPrompt = prompt.replace("{{topic}}", topic)

    for model in llm_models:
        payload = {
            "workspaceID": "c38690ce92ba0d88e11b856d",
            "definition": "",
            "analyticsName": "",
            "prompt": finalPrompt,
        }

        try:
            # Send API request
            response = requests.post(url=LLUMO_ENDPOINT, json=payload, headers=headers, timeout=18)
            response_json = response.json()
            llm_output = json.loads(response_json["data"])



            # Append data to DataFrame correctly
            data.loc[indx, "Prompt"] = finalPrompt
            data.loc[indx, f"output_{model}"] = llm_output['output']

            print(f"✅ Successfully fetched response from {model} for the given prompt.")

        except Exception as e:
            print(f"❌ Error fetching response from {model} for the given prompt: {e}")
            continue
```

```python
# final dataframe
print(data)
```
