# 📊 Evaluating LLM Outputs using LLUMO 


This notebook evaluates the quality of multiple LLM-generated responses based on key metrics such as **Response Correctness**, **Hallucination**, and **Input Bias**, among others. The evaluation is performed using the **LLUMO SDK**.

## 🧠 Importing LLUMO Client

To get started with LLUMO, import the core client and model constants:


```python
from llumo import  LlumoClient,AVAILABLEMODELS
```

## 🔑 Initializing LLUMO Client

To begin using LLUMO, initialize the `LlumoClient` with your API key:


```python

client = LlumoClient(api_key="Enter Your LLumo API KEY")
```

```python

dataset = {
    "topic": ["Quantum Physics", "Politics"]  
}

models = [
    AVAILABLEMODELS.CHAT_BISON,
    AVAILABLEMODELS.GOOGLE_15_FLASH,    
]
api_key = "Enter your Google Api Key"
prompt_template=["Explain {{topic}} to me like I'm a 5 year old. in 5 words"]
res = client.run_sweep(templates=prompt_template, dataset=dataset, model_aliases=models, apiKey=api_key, evals=["Response Correctness"])


print(res.to_csv("data.csv",index=False))
```
