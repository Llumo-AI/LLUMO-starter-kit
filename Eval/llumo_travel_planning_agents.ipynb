```python
# Importing necessary libraries
import pandas as pd

# Llumo package imports
import llumo
from llumo import LlumoClient
from llumo.functionCalling import LlumoAgent
```

# ⚛ Defining Tools

```python
# 🌍 Suggest Destinations Function
def suggestDestinations(season, continent):
    destinations = {
        "summer": {
            "Europe": ["Barcelona", "Nice", "Santorini"],
            "Asia": ["Bali", "Phuket", "Goa"],
            "North America": ["Miami", "Cancun", "Los Angeles"]
        },
        "winter": {
            "Europe": ["Vienna", "Prague", "Zermatt"],
            "Asia": ["Tokyo", "Seoul", "Dubai"],
            "North America": ["New York", "Toronto", "Aspen"]
        }
    }
    return {"suggestions": destinations.get(season.lower(), {}).get(continent, ["No suggestions available"])}

# 🏨 Hotel Information Function
def hotelInfo(city):
    hotels = {
        "Barcelona": {"name": "Hotel Arts", "rating": "5 stars", "price": "$300/night"},
        "Bali": {"name": "The Mulia", "rating": "5 stars", "price": "$250/night"},
        "Tokyo": {"name": "Park Hyatt", "rating": "5 stars", "price": "$400/night"},
        "Miami": {"name": "Fontainebleau", "rating": "4 stars", "price": "$350/night"},
        "Dubai": {"name": "Atlantis The Palm", "rating": "5 stars", "price": "$450/night"}
    }
    return hotels.get(city, {"message": "No hotel info available for this city"})

# 🎒 Packing List Recommendation Function
def packingList(season, activityType):
    items = {
        "beach": ["Sunscreen", "Swimsuit", "Sunglasses", "Flip-flops"],
        "hiking": ["Backpack", "Hiking Boots", "Water Bottle", "First Aid Kit"],
        "city tour": ["Comfortable Shoes", "Power Bank", "Light Jacket", "Camera"]
    }
    basic_items = {
        "summer": ["Hat", "Light clothes", "Water bottle"],
        "winter": ["Gloves", "Warm clothes", "Thermal bottle"]
    }
    return {"essentials": basic_items.get(season.lower(), []) + items.get(activityType.lower(), [])}

# ✈️ Flight Booking Info Function (Dummy)
def flightBookingInfo(destination, date):
    return {
        "destination": destination,
        "date": date,
        "available_flights": [
            {"airline": "AirwaysX", "price": "$400", "departure": "10:00 AM"},
            {"airline": "FlyHigh", "price": "$450", "departure": "3:00 PM"}
        ]
    }
```

# 📚Agent Registration


```python
destinationAgent = LlumoAgent(
    name="suggestDestinations",
    description="Suggests travel destinations based on season and continent",
    parameters={
        "season": {"type": "string", "description": "Season of travel like summer or winter"},
        "continent": {"type": "string", "description": "Continent for travel, e.g., Europe, Asia"}
    },
    function=suggestDestinations
)

hotelAgent = LlumoAgent(
    name="hotelInfo",
    description="Provides hotel information for a given city",
    parameters={
        "city": {"type": "string", "description": "City for which hotel info is needed"}
    },
    function=hotelInfo
)

packingAgent = LlumoAgent(
    name="packingList",
    description="Suggests packing list based on season and activity type",
    parameters={
        "season": {"type": "string", "description": "Season like summer or winter"},
        "activityType": {"type": "string", "description": "Type of travel activity like beach, hiking, etc."}
    },
    function=packingList
)

flightAgent = LlumoAgent(
    name="flightBookingInfo",
    description="Provides dummy flight booking options for a destination and date",
    parameters={
        "destination": {"type": "string", "description": "Destination city"},
        "date": {"type": "string", "description": "Date of travel"}
    },
    function=flightBookingInfo
)

print("✅ All Travel Agents created successfully")
```

# 📝 Dummy Queries for Evaluation

```python
df = pd.DataFrame({
    "query": [
        "Where should I travel in summer in Europe?",
        "Suggest a hotel in Bali",
        "I am planning a winter trip to Asia, any ideas?",
        "What should I pack for a hiking trip in summer?",
        "I need to know the best hotel in Tokyo",
        "List flights available to Dubai on 25th December",
        "Can you recommend cities for a winter trip in Europe?",
        "Give me some packing essentials for a city tour during winter",
        "Book a flight to Miami for tomorrow",
        "Help me find a hotel in Barcelona"
    ]
})
```

🔐 Keys Setup & Output Generation
- Setting up the Open AI key
- Setting up Llumo Client
- Output Generation & Llumo Evaluation


```python
# API Keys
openaiKey = "Open AI Key Here"   # Replace with your OpenAI key
client = LlumoClient(api_key="Enter your Llumo API Key")
# List of registered agents
agents = [destinationAgent, hotelAgent, packingAgent, flightAgent]

print("🚀 Starting Output Generation & Evaluation...")
results = client.evaluateAgents(df, model="openai", agents=agents, model_api_key=openaiKey)

# Display results
result_df = results

```
